```python
crops=crops.dropna(subset=['YIELD_KG'])
```

```python
crops['YIELD_KG'].isna().sum()
```

```python
crops['AREA_ACRES']=crops['AREA_ACRES'].fillna(crops['AREA_ACRES'].mean())
```

```python
crops['AREA_ACRES'].isna().sum()
```

```python
crops['DISTRICT']=crops['DISTRICT'].fillna('Unknown')
crops['DISTRICT'].isna().sum()
```

```python
crops['FERTILIZER_USED']=crops['FERTILIZER_USED'].fillna('No')
crops['FERTILIZER_USED'].isna().sum()
```

```python
crops['YIELD_VALUE']=crops['YIELD_VALUE'].fillna(crops["YIELD_VALUE"].mean())
crops['YIELD_VALUE'].isna().sum()
```

```python
crops['HARVEST_DATE']=pd.to_datetime(crops['HARVEST_DATE'],errors='coerce')
```

```python
crops.info()
```

```python
crops['FARM_ID']=crops['FARM_ID'].str.strip("F")
```

```python
crops.head(2)
```

```python
crops['FARM_ID']=pd.to_numeric(crops['FARM_ID'],errors='coerce')
```

```python
crops.info()
```

```python
yield_sum=crops.groupby('CROP_TYPE')['YIELD_KG'].sum()
yield_sum
```

```python
import matplotlib.pyplot as plt
```

```python
x=yield_sum.index
y=yield_sum.values
plt.bar(x,y)
plt.xlabel('Crop Type')
plt.ylabel('Total Yield(KG)')
plt.title(' Total Yield by Crop Type (KG)')
plt.show()
```

```python
table=crops.pivot_table(index='HARVEST_DATE',values='YIELD_KG',aggfunc='sum')
table
```

```python
x=table.index
y=table.values
plt.figure(figsize=(15,5))
plt.title('Yield Trend Over Time(KG)')
plt.ylabel('Yield(KG)')
plt.plot(x,y)
```
